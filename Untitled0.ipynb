```python
import random
K = 2
for _ in range(K):
    n = [random.randint(1, 9) for _ in range(random.randint(2, 5))]
    if n == sorted(n) and len(set(n)) == len(n):
        print(1)
    elif n == sorted(n, reverse=True) and len(set(n)) == len(n):
        print(-1)
    else:
        print(0)
```
